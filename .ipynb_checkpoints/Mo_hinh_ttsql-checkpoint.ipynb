```python
import os
import re
import sqlite3
import torch
import pandas as pd
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel
from transformers import AutoModelForCausalLM, AutoTokenizer

app = FastAPI(title="Text-to-SQL Warehouse API", version="1.0")

# ==========================================
# CẤU HÌNH ĐƯỜNG DẪN & MÔ HÌNH TRÊN Ổ D
# ==========================================
# Kết hợp thư mục gốc và tên file cơ sở dữ liệu SQLite
DB_DIR = r"D:\đại học quy nhơn\thực hành chuyên đề\phat_trien_he_thong_ql_kho"

# 1. Sửa lại đường dẫn chính xác (bỏ thư mục "csv" nếu file nằm trực tiếp ở thư mục gốc)
DB_PATH = os.path.join(DB_DIR, "quan_ly_kho.db")

CACHE_DIR = r"D:\huggingface_models"
MODEL_NAME = "Qwen/Qwen2.5-Coder-1.5B-Instruct"

print("Đang tải mô hình Text-to-SQL từ ổ D...")
tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME, 
    cache_dir=CACHE_DIR
)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME, 
    cache_dir=CACHE_DIR,
    torch_dtype=torch.float16, 
    device_map="auto"
)
print("Tải mô hình thành công!")

# ==========================================
# SCHEMA CƠ SỞ DỮ LIỆU (9 BẢNG)
# ==========================================
database_schema = """
CREATE TABLE NhanVien (
    MaNV TEXT PRIMARY KEY,
    HoTen TEXT,
    ChucVu TEXT
);

CREATE TABLE NhaCungCap (
    MaNCC TEXT PRIMARY KEY,
    TenNCC TEXT,
    DiaChi TEXT,
    DienThoai TEXT
);

CREATE TABLE VatTu (
    MaVT TEXT PRIMARY KEY,
    TenVT TEXT,
    DonViTinh TEXT,
    GiaGoc REAL
);

CREATE TABLE Kho (
    MaKho TEXT PRIMARY KEY,
    TenKho TEXT,
    DiaChi TEXT
);

CREATE TABLE TonKho (
    MaKho TEXT,
    MaVT TEXT,
    SoLuongTon INTEGER,
    PRIMARY KEY (MaKho, MaVT)
);

CREATE TABLE PhieuNhap (
    MaPN TEXT PRIMARY KEY,
    NgayNhap TEXT,
    MaNCC TEXT,
    MaNV TEXT
);

CREATE TABLE ChiTietPhieuNhap (
    MaPN TEXT,
    MaVT TEXT,
    SoLuong INTEGER,
    DonGia REAL,
    PRIMARY KEY (MaPN, MaVT)
);

CREATE TABLE PhieuXuat (
    MaPX TEXT PRIMARY KEY,
    NgayXuat TEXT,
    NguoiNhan TEXT,
    MaNV TEXT
);

CREATE TABLE ChiTietPhieuXuat (
    MaPX TEXT,
    MaVT TEXT,
    SoLuong INTEGER,
    DonGia REAL,
    PRIMARY KEY (MaPX, MaVT)
);
"""

# ==========================================
# ĐỊNH NGHĨA API REQUEST MODEL
# ==========================================
class QueryRequest(BaseModel):
    user_query: str

# ==========================================
# HÀM XỬ LÝ TEXT-TO-SQL & THỰC THI
# ==========================================
# 2. Hàm xử lý Text-to-SQL đã được tối ưu bộ lọc
def generate_and_execute_sql(user_query: str):
    # 1. Xây dựng prompt chuẩn ChatML của Qwen
    prompt = f"""<|im_start|>system
Bạn là một công cụ dịch Text-to-SQL cho SQLite. Chỉ xuất ra câu lệnh SQL hợp lệ dựa vào schema sau:
{database_schema}
<|im_end|>
<|im_start|>user
{user_query}<|im_end|>
<|im_start|>assistant
SELECT """

    # 2. Tokenize và sinh câu lệnh từ model
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    outputs = model.generate(
        **inputs, 
        max_new_tokens=150, 
        temperature=0.01, 
        do_sample=False
    )
    
    raw_output = tokenizer.decode(outputs[0], skip_special_tokens=True)
    
    # 3. Ghép lại chữ SELECT vì ta đã cắt bớt ở prompt, sau đó dùng Regex để lấy trọn vẹn câu lệnh SQL
    full_text = "SELECT " + raw_output.split("SELECT")[-1] if "SELECT" in raw_output else raw_output
    
    # Lọc lấy từ SELECT cho đến dấu chấm phẩy kết thúc
    match = re.search(r"(SELECT\s+.*?;)", full_text, re.IGNORECASE | re.DOTALL)
    if match:
        sql_query = match.group(1)
    else:
        sql_query = full_text.split("\n")[0].strip() # Lấy dòng đầu tiên nếu không thấy dấu chấm phẩy

    # 4. Thực thi câu lệnh SQL vào SQLite
    try:
        conn = sqlite3.connect(DB_PATH)
        cursor = conn.cursor()
        cursor.execute(sql_query)
        rows = cursor.fetchall()
        conn.close()

        return {
            "status": "success",
            "sql_query": sql_query,
            "data": rows
        }
    except Exception as e:
        return {
            "status": "error",
            "sql_query": sql_query,
            "message": str(e)
        }

# ==========================================
# API ENDPOINT CHO FASTAPI
# ==========================================
@app.get("/api/chat")
def text_to_sql_endpoint(prompt: str):
    prompt_lower = prompt.lower().strip()
    
    if any(w in prompt_lower for w in ["chào", "hi", "hello", "bạn là ai"]):
        return {"status": "success", "reply": "👋 Chào bạn, tôi là trợ lý AI Pro quản lý kho VLXD."}

    try:
        input_text = f"""You are an expert SQLite SQL generator. 
Given the database schema, write a valid SQLite SQL query for the question. Return ONLY the raw SQL query starting with SELECT, ending with a semicolon (;). No explanation.

Schema:
{DB_SCHEMA}

Question: {prompt}
SQL: SELECT"""

        inputs = tokenizer(input_text, return_tensors="pt").to(model.device)
        outputs = model.generate(
        **inputs, 
        max_new_tokens=150, 
        do_sample=False
    )
        
        raw_output = tokenizer.decode(outputs[0], skip_special_tokens=True)
        
        # Tách lấy đoạn SQL bằng biểu thức chính quy (Regex) tìm từ chữ SELECT đến dấu chấm phẩy
        match = re.search(r"(SELECT\s+.*?;)", raw_output, re.IGNORECASE | re.DOTALL)
        if match:
            sql_query = match.group(1)
        else:
            # Fallback nếu không tìm thấy dấu chấm phẩy
            sql_query = "SELECT " + raw_output.split("SQL:")[-1].strip()
            if "SELECT" in sql_query:
                sql_query = sql_query[sql_query.find("SELECT"):]

        # Thực thi SQL lên SQLite
        conn = sqlite3.connect(DB_PATH)
        cursor = conn.cursor()
        cursor.execute(sql_query)
        rows = cursor.fetchall()
        conn.close()

        if not rows:
            return {"status": "success", "reply": f"📊 Câu lệnh SQL sinh ra: `{sql_query}`\n\nKhông tìm thấy dữ liệu phù hợp."}
            
        result_str = f"🔍 **SQL:** `{sql_query}`\n\n📊 **Kết quả:**\n"
        for row in rows:
            result_str += f"- {row}\n"
            
        return {"status": "success", "reply": result_str}

    except Exception as e:
        return {"status": "error", "reply": f"⚠️ Lỗi thực thi SQL: {str(e)} \n- Câu lệnh gây lỗi: {sql_query if 'sql_query' in locals() else 'N/A'}"}
```

```python
generate_and_execute_sql("Liệt kê toàn bộ danh sách nhân viên trong kho")
```
