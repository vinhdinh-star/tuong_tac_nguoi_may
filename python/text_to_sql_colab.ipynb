````python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
import sqlite3
import pandas as pd

# ==========================================
# BƯỚC 1: KHỞI TẠO VÀ NẠP DỮ LIỆU MẪU VÀO SQLite
# ==========================================
conn = sqlite3.connect('quan_ly_kho.db', check_same_thread=False)
cursor = conn.cursor()

# Tạo toàn bộ các bảng theo chuẩn sơ đồ cơ sở dữ liệu kho
cursor.executescript('''
CREATE TABLE IF NOT EXISTS NhanVien (
    MaNV TEXT PRIMARY KEY,
    HoTen TEXT,
    ChucVu TEXT
);

CREATE TABLE IF NOT EXISTS NhaCungCap (
    MaNCC TEXT PRIMARY KEY,
    TenNCC TEXT,
    DiaChi TEXT,
    DienThoai TEXT
);

CREATE TABLE IF NOT EXISTS VatTu (
    MaVT TEXT PRIMARY KEY,
    TenVT TEXT,
    DonViTinh TEXT,
    GiaGoc REAL
);

CREATE TABLE IF NOT EXISTS Kho (
    MaKho TEXT PRIMARY KEY,
    TenKho TEXT,
    DiaChi TEXT
);

CREATE TABLE IF NOT EXISTS TonKho (
    MaKho TEXT,
    MaVT TEXT,
    SoLuongTon INTEGER,
    PRIMARY KEY (MaKho, MaVT)
);

CREATE TABLE IF NOT EXISTS PhieuNhap (
    MaPN TEXT PRIMARY KEY,
    NgayNhap TEXT,
    MaNCC TEXT,
    MaNV TEXT
);

CREATE TABLE IF NOT EXISTS ChiTietPhieuNhap (
    MaPN TEXT,
    MaVT TEXT,
    SoLuong INTEGER,
    DonGia REAL,
    PRIMARY KEY (MaPN, MaVT)
);

CREATE TABLE IF NOT EXISTS PhieuXuat (
    MaPX TEXT PRIMARY KEY,
    NgayXuat TEXT,
    NguoiNhan TEXT,
    MaNV TEXT
);

CREATE TABLE IF NOT EXISTS ChiTietPhieuXuat (
    MaPX TEXT,
    MaVT TEXT,
    SoLuong INTEGER,
    DonGia REAL,
    PRIMARY KEY (MaPX, MaVT)
);
''')

# Thêm dữ liệu mẫu chuẩn xác
cursor.executescript('''
INSERT OR IGNORE INTO NhanVien VALUES ('NV01', 'Nguyen Van A', 'Thu kho');
INSERT OR IGNORE INTO NhanVien VALUES ('NV02', 'Tran Thi B', 'Nhan vien ban hang');

INSERT OR IGNORE INTO NhaCungCap VALUES ('NCC01', 'Cong ty VLXD Mien Trung', 'Quy Nhon', '0351234567');
INSERT OR IGNORE INTO NhaCungCap VALUES ('NCC02', 'Dai ly Sat Thep Nam A', 'Da Nang', '0905111222');

INSERT OR IGNORE INTO VatTu VALUES ('VT01', 'Xi mang Hoang Thach', 'Bao', 85000);
INSERT OR IGNORE INTO VatTu VALUES ('VT02', 'Sat phi 10 Pomina', 'Cay', 115000);
INSERT OR IGNORE INTO VatTu VALUES ('VT03', 'Gach ong 2 lo', 'Vien', 1200);

INSERT OR IGNORE INTO Kho VALUES ('K01', 'Kho Tong Quy Nhon', '150 An Duong Vuong');
INSERT OR IGNORE INTO Kho VALUES ('K02', 'Kho Phu', '12 Tran Hung Dao');

INSERT OR IGNORE INTO TonKho VALUES ('K01', 'VT01', 500);
INSERT OR IGNORE INTO TonKho VALUES ('K01', 'VT02', 200);
INSERT OR IGNORE INTO TonKho VALUES ('K02', 'VT03', 1000);

INSERT OR IGNORE INTO PhieuNhap VALUES ('PN01', '2026-06-01', 'NCC01', 'NV01');
INSERT OR IGNORE INTO PhieuNhap VALUES ('PN02', '2026-06-05', 'NCC02', 'NV01');

INSERT OR IGNORE INTO ChiTietPhieuNhap VALUES ('PN01', 'VT01', 500, 80000);
INSERT OR IGNORE INTO ChiTietPhieuNhap VALUES ('PN02', 'VT02', 200, 110000);

INSERT OR IGNORE INTO PhieuXuat VALUES ('PX01', '2026-06-10', 'Cong ty Xay dung ABC', 'NV02');
INSERT OR IGNORE INTO ChiTietPhieuXuat VALUES ('PX01', 'VT01', 50, 90000);
''')
conn.commit()
print("Đã khởi tạo database và dữ liệu mẫu thành công!")

# ==========================================
# BƯỚC 2: TẢI MODEL AI (SQLCoder-7B)
# ==========================================
model_id = "defog/sqlcoder-7b-2"
print("\nĐang tải Tokenizer và Model (quá trình này mất vài phút)...")

tokenizer = AutoTokenizer.from_pretrained(model_id)

quantization_config = BitsAndBytesConfig(
    load_in_8bit=True,
    llm_int8_enable_fp32_cpu_offload=True
)

model = AutoModelForCausalLM.from_pretrained(
    model_id,
    quantization_config=quantization_config,
    device_map="auto"
)
print("Đã tải xong Model!")

# Khai báo Schema tường minh cho AI hiểu cấu trúc bảng
database_schema = """
Cơ sở dữ liệu SQLite Quản lý Kho gồm các bảng sau:
1. NhanVien (MaNV, HoTen, ChucVu)
2. NhaCungCap (MaNCC, TenNCC, DiaChi, DienThoai)
3. VatTu (MaVT, TenVT, DonViTinh, GiaGoc)
4. Kho (MaKho, TenKho, DiaChi)
5. TonKho (MaKho, MaVT, SoLuongTon)
6. PhieuNhap (MaPN, NgayNhap, MaNCC, MaNV)
7. ChiTietPhieuNhap (MaPN, MaVT, SoLuong, DonGia)
8. PhieuXuat (MaPX, NgayXuat, NguoiNhan, MaNV)
9. ChiTietPhieuXuat (MaPX, MaVT, SoLuong, DonGia)
"""

# ==========================================
# BƯỚC 3: HÀM TẠO SQL BẰNG AI VÀ THỰC THI
# ==========================================
def generate_and_execute_sql(user_query: str):
    prompt = f"""[INST] Bạn là một chuyên gia Text-to-SQL chuẩn SQLite. Dựa vào schema sau:
{database_schema}
Hãy viết câu lệnh SQL SQLite chính xác cho câu hỏi bằng tiếng Việt sau: "{user_query}"
Lưu ý: Chỉ trả về duy nhất câu lệnh SQL thuần túy, không kèm giải thích hay định dạng markdown. [/INST]"""

    inputs = tokenizer(prompt, return_tensors="pt").to("cuda")
    torch.cuda.empty_cache()

    outputs = model.generate(
        **inputs,
        max_new_tokens=150,
        temperature=0.1,
        do_sample=False
    )
    sql_query = tokenizer.decode(outputs[0], skip_special_tokens=True).strip()

    # Xử lý cắt bỏ phần prompt nếu mô hình sinh lặp lại
    if "[/INST]" in sql_query:
        sql_query = sql_query.split("[/INST]")[-1].strip()

    # Làm sạch các thẻ markdown nếu có
    sql_query = sql_query.replace("```sql", "").replace("```", "").strip()

    print(f"\n--- CÂU HỎI: {user_query} ---")
    print(f"CÂU LỆNH SQL ĐƯỢC TẠO: {sql_query}")

    # Thực thi SQL trên SQLite
    try:
        df_result = pd.read_sql_query(sql_query, conn)
        print("KẾT QUẢ DỮ LIỆU:")
        display(df_result)
    except Exception as e:
        print(f"LỖI THỰC THI SQL: {e}")

# ==========================================
# BƯỚC 4: KIỂM THỬ TRỰC TIẾP
# ==========================================
generate_and_execute_sql("Cho tôi danh sách vật tư tồn kho hiện tại")
generate_and_execute_sql("Nhà cung cấp nào đang có trong hệ thống?")
generate_and_execute_sql("Thống kê danh sách nhân viên")
````

````python
import sqlite3
import pandas as pd

# 1. Đọc file CSV (tự động nhận diện dấu phân tách tab hay phẩy)
try:
    df_nv = pd.read_csv('NhanVien.csv', sep=None, engine='python')
except Exception as e:
    # Dự phòng nếu file dùng tab rõ ràng
    df_nv = pd.read_csv('NhanVien.csv', sep='\t')

# 2. Chuyển toàn bộ tên cột thành chữ thường (ví dụ: MaNV -> manv)
df_nv.columns = [c.lower().strip() for c in df_nv.columns]

print("Tên các cột sau khi chuẩn hóa:", df_nv.columns.tolist())

# 3. Kết nối SQLite và tạo bảng nhan_vien chuẩn chữ thường
conn = sqlite3.connect('quan_ly_kho.db', check_same_thread=False)
cursor = conn.cursor()

cursor.executescript('''
DROP TABLE IF EXISTS nhan_vien;
CREATE TABLE nhan_vien (
    manv TEXT PRIMARY KEY,
    hoten TEXT,
    chucvu TEXT
);
''')

# 4. Nạp dữ liệu vào bảng
df_nv.to_sql('nhan_vien', conn, if_exists='append', index=False)
conn.commit()
conn.close()

print(f"Thành công! Đã nạp {len(df_nv)} nhân viên vào bảng nhan_vien.")
````
